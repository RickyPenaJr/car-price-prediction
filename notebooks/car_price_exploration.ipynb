# Car Price Exploration
Exploratory Data Analysis and visualizations for car price dataset.

```python
import pandas as pd
from src.data_processing import load_data

df = load_data()
df.head()
```

```python
import matplotlib.pyplot as plt

# Price Distribution
plt.figure(figsize=(6,4))
df['price'].hist(bins=40)
plt.xlabel('Price')
plt.ylabel('Count')
plt.title('Price Distribution')
plt.show()
```

```python
# Price vs. Mileage
plt.figure(figsize=(6,4))
plt.scatter(df['mileage'], df['price'], alpha=0.5)
plt.xlabel('Mileage')
plt.ylabel('Price')
plt.title('Price vs. Mileage')
plt.show()
```
